```python
quantidade = 0
positivos = 0
negativos = 0
soma = 0

while True:
    num = int(input("Digite um número: "))
    if num == 0:
        break

    quantidade += 1
    soma += num

    if num > 0:
        positivos += 1
    else:
        negativos += 1

print(f"Quantidade: {quantidade}")
print(f"Positivos: {positivos}")
print(f"Negativos: {negativos}")
print(f"Soma: {soma}")
```

```python
numero_secreto = 7

palpite = int(input("Tente adivinhar o número secreto: "))

while palpite != numero_secreto:
    if palpite < numero_secreto:
        print("Tente um número maior")
    else:
        print("Tente um número menor")
    palpite = int(input("Tente novamente: "))

print("Parabéns! Você acertou.")
```

```python
senha_correta = "1234"
senha = input("Digite a senha: ")

while senha != senha_correta:
    senha = input("Senha incorreta. Tente novamente: ")

print("Acesso permitido!")
```

```python
num = int(input("Digite um número: "))

for i in range(1, 11):
    print(f"{num} x {i} = {num * i}")
```

```python
maior = None

for i in range(6):
    num = int(input(f"Digite o {i+1}º número: "))
    if maior is None or num > maior:
        maior = num

print(f"Maior valor: {maior}")
```

```python
soma_positivos = 0

for i in range(8):
    num = int(input(f"Digite o {i+1}º número: "))
    if num > 0:
        soma_positivos += num

print(f"Soma dos positivos: {soma_positivos}")
```

```python
pares = 0

for i in range(10):
    num = int(input(f"Digite o {i+1}º número: "))
    if num % 2 == 0:
        pares += 1

print(f"Foram digitados {pares} números pares.")
```

```python
N = int(input("Digite N: "))

soma = 0
for i in range(1, N + 1):
    soma += i

print(f"Soma: {soma}")
```

```python
# Leitura dos três números inteiros
num1 = int(input("Primeiro número: "))
num2 = int(input("Segundo número: "))
num3 = int(input("Terceiro número: "))

# Comparação direta usando estruturas condicionais
if num1 >= num2 and num1 >= num3:
    maior = num1
elif num2 >= num1 and num2 >= num3:
    maior = num2
else:
    maior = num3

print(f"Maior número: {maior}")
```

```python
nota1 = float(input("Digite a primeira nota: "))
nota2 = float(input("Digite a segunda nota: "))

media = (nota1 + nota2) / 2

if media >= 7:
    situacao = "Aprovado"
elif media >= 5:
    situacao = "Recuperação"
else:
    situacao = "Reprovado"

print(f"Média: {media:.2f}")
print(f"Situação: {situacao}")
```
